## Побудова фракталів з використанням ітерованих функцій

```python
import matplotlib.pyplot as plt
import random
from math import cos, sin
```

```python
def draw_fractal(transformations, probabilities, title, n_points=10000):
    x, y = 0.0, 0.0
    X, Y = [], []

    for _ in range(n_points):
        r = random.random()
        total = 0
        for i, p in enumerate(probabilities):
            total += p
            if r <= total:
                a, b, c, d, e, f = transformations[i]
                x_new = a * x + b * y + e
                y_new = c * x + d * y + f
                x, y = x_new, y_new
                break
        X.append(x)
        Y.append(y)

    plt.scatter(X, Y, s=0.1, color='green')
    plt.axis('equal')
    plt.title(title)
    plt.show()
```

```python
transforms = [
    (0.1400, 0.0100, 0.0000, 0.5100, -0.0800, -1.3100),
    (0.4300, 0.5200, -0.4500, 0.5000, 1.4900, -0.7500),
    (0.4500, -0.4900, 0.4700, 0.4700, -1.6200, -0.7400),
    (0.4900, 0.0000, 0.0000, 0.5100, 0.0200, 1.6200)
]
probs = [0.25, 0.25, 0.25, 0.25]

draw_fractal(transforms, probs, title='Кленовий лист')
```
