```python
import turtle
import time
```

```python
def apply_rules(axiom, rules, iterations):
    for _ in range(iterations):
        axiom = ''.join(rules.get(c, c) for c in axiom)
    return axiom
```

```python
def draw_l_system(instructions, angle, length, delay=0.01):
    for cmd in instructions:
        if cmd == "F":
            turtle.forward(length)
            time.sleep(delay)
        elif cmd == "-":
            turtle.right(angle)
        elif cmd == "+":
            turtle.left(angle)
```

```python
def setup_turtle():
    turtle.reset()
    turtle.speed(1000)
    turtle.penup()
    turtle.goto(-200, 0)
    turtle.pendown()
```

```python
def main():
    fractals = {
        "1": ("Фрактальна дошка", "F+F+F+F", {"F": "FF+F+F+F+FF"}, 90),
        "2": ("Сніжинка Коха", "F++F++F", {"F": "F-F++F-F"}, 60),
        "3": ("Крива Леві", "F", {"F": "-F++F-"}, 45),
        "4": ("Крива дракона", "FX", {"X": "X+YF+", "Y": "-FX-Y"}, 90),
        "5": ("Квадратний острівець Коха 2", "F+F+F+F", {"F": "F-FF+FF+F+F-F-FF+F+F-F-FF-FF+F"}, 90),
        "6": ("Фрактальна зірочка", "X+X+X+X+X+X+X+X", {"X": "X+YF++YF-FX--FXFX-YF+X", "Y": "-FX+YFYF++YF+FX--FX-YF"}, 45),
        "7": ("П’ятикутна фрактальна сніжинка", "F++F++F++F++F", {"F": "F++F++F+++++F-F++F"}, 36),
        "8": ("Фрактальні кільця", "XF", {"X": "X+YF++YF-FX--FXFX-YF+", "Y": "-FX+YFYF++YF+FX--FX-Y"}, 60)
    }

    print("Оберіть фрактал для побудови:")
    for key, (name, *_rest) in fractals.items():
        print(f"{key}. {name}")

    choice = input("Ваш вибір (1-8): ").strip()
    iterations = int(input("Кількість ітерацій: "))

    if choice not in fractals:
        raise ValueError

    name, axiom, rules, angle = fractals[choice]
    instructions = apply_rules(axiom, rules, iterations)

    setup_turtle()
    draw_l_system(instructions, angle, length=5, delay=0.01)
    turtle.done()
```

```python
if __name__ == "__main__":
    main()
```
