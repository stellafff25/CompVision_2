# Обчислення фрактальної розмірності та побудова фрактала

Ми розглядаємо фрактал на площині `[0, 1] × [0, 1]`, утворений з точок `(x, y)`, де в **десятковому представленні** чисел **x** і **y** **не використовуються цифри 5 і 9**.

---

## Теоретична частина

Кожне число в інтервалі `[0;1]` можна представити як нескінченний десятковий дріб:

x = 0.a₁a₂a₃..., де aᵢ ∈ {0,1,2,3,4,6,7,8}

### Загальна формула:

D = log(N) / log(1/r)


У нашому випадку:

- `N = 8` — кількість допустимих цифр
- `r = 1/10` — масштаб при кожному поділі

Тоді:

D = log(8) / log(10) ≈ 0.903

```python
import matplotlib.pyplot as plt
import random
```

```python
def is_valid_decimal(n, max_digits=5):
    s = f"{n:.{max_digits}f}"[2:]
    return all(c not in '59' for c in s)
```

```python
def generate_fractal_points(n_points=100000, max_digits=5):
    points = []
    while len(points) < n_points:
        x = random.random()
        y = random.random()
        if is_valid_decimal(x, max_digits) and is_valid_decimal(y, max_digits):
            points.append((x, y))
    return points
```

```python
points = generate_fractal_points()
x_vals, y_vals = zip(*points)

plt.figure(figsize=(8, 8))
plt.scatter(x_vals, y_vals, s=0.1, color='black')
plt.xlabel("x")
plt.ylabel("y")
plt.axis('square')
plt.grid(False)
plt.show()
```
